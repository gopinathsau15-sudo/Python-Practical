```python
num = int(input("Enter an integer between 1 and 9999: "))

if num < 1 or num > 9999:
    print("Please enter a number between 1 and 9999.")

else:
    # Check Even or Odd
    if num % 2 == 0:
        print(num, "is Even")
    else:
        print(num, "is Odd")

    # Check Palindrome
    if str(num) == str(num)[::-1]:
        print(num, "is a Palindrome")
    else:
        print(num, "is not a Palindrome")

    # Check divisibility by sum of digits
    digit_sum = sum(int(digit) for digit in str(num))

    if num % digit_sum == 0:
        print(num, "is divisible by the sum of its digits")
    else:
        print(num, "is not divisible by the sum of its digits")
```

```python
Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])

    i += 1

print("Animals with 7 letters:")
print(seven_letter_animals)
```
