```python
import pandas as pd

data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Arjun',
             'Neha', 'Karan', 'Simran', 'Rohit', 'Anjali'],
    'Age': [25, 32, 28, 35, 22, 40, 31, 27, 29, 33],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [2000, 3500, 1500, 4000, 2500,
                        5000, 3000, 4500, 1800, 3200]
}

df = pd.DataFrame(data)

# Display first 5 rows
print("First 5 Rows:")
print(df.head())

# Summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].median()
)

print("\nData After Handling Missing Values:")
print(df)
```

```python
print(df[df['Age'] > 30])
```
