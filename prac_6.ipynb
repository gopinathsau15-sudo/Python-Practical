```python
class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def Display(self):
        print("Name:", self.name)
        print("ID:", self.id)


class Emp(Person):
    def __init__(self, name, id):
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")


# Create an object of Emp
Emp_details = Emp("Mayank", 103)

# Display details
Emp_details.Display()
Emp_details.Print()
```

```python
word1 = input("Enter first word: ")
word2 = input("Enter second word: ")

if sorted(word1.lower()) == sorted(word2.lower()):
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

```python
number = int(input("Enter a number: "))

digits = str(number)
power = len(digits)

sum_of_powers = 0

for digit in digits:
    sum_of_powers += int(digit) ** power

if sum_of_powers == number:
    print(number, "is an Armstrong number.")
else:
    print(number, "is not an Armstrong number.")
```

```python
class Vehicle:
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.color = "white"
        self.seating_capacity = 0

    # Assign seating capacity
    def assign_seating_capacity(self, capacity):
        self.seating_capacity = capacity

    # Display vehicle properties
    def display_properties(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)


# Create two objects
car1 = Vehicle(200, 20)
car1.assign_seating_capacity(5)

car2 = Vehicle(180, 25)
car2.assign_seating_capacity(4)

# Display first vehicle
print("First Vehicle:")
car1.display_properties()

print()

# Display second vehicle
print("Second Vehicle:")
car2.display_properties()
```
